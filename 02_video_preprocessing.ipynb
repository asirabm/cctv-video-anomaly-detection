**Extract the videos from zip file **

```python
import zipfile
import os

extract_path = "/content/UCF_Crime/videos"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Extraction completed.")
print("Videos extracted to:", extract_path)
```

**Check the extracted videos**

```python
import os

video_root = "/content/UCF_Crime/videos/Anomaly-Videos-Part-1"

extracted_videos = []

for root, dirs, filenames in os.walk(video_root):
    for filename in filenames:
        if filename.lower().endswith(".mp4"):
            extracted_videos.append(
                os.path.join(root, filename)
            )

print("Extracted MP4 videos:", len(extracted_videos))

print("\nFirst 10 videos:")
for video in extracted_videos[:10]:
    print(video)
```

**Count videos in each category**

```python
from collections import Counter
import os

category_counts = Counter()

for video in extracted_videos:
    category = os.path.basename(os.path.dirname(video))
    category_counts[category] += 1

print("Videos in each category:")

for category, count in sorted(category_counts.items()):
    print(f"{category}: {count}")

print("Total videos:", sum(category_counts.values()))
```

**Download the official split files**

```python
from huggingface_hub import hf_hub_download

repo_id = "jinmang2/ucf_crime"
split_folder = "UCF_Crimes-Train-Test-Split/Anomaly_Detection_splits/"

train_split_path = hf_hub_download(
    repo_id=repo_id,
    filename=split_folder + "Anomaly_Train.txt",
    repo_type="dataset"
)

test_split_path = hf_hub_download(
    repo_id=repo_id,
    filename=split_folder + "Anomaly_Test.txt",
    repo_type="dataset"
)

print("Training split:", train_split_path)
print("Testing split:", test_split_path)
```

**Identify training and testing videos**

```python

with open(train_split_path, "r") as f:
    train_lines = [line.strip() for line in f if line.strip()]

with open(test_split_path, "r") as f:
    test_lines = [line.strip() for line in f if line.strip()]


train_names = {line.split("/")[-1].replace("\\", "/") for line in train_lines}
test_names = {line.split("/")[-1].replace("\\", "/") for line in test_lines}


train_videos = []
test_videos = []
unmatched_videos = []

for video in extracted_videos:
    video_name = os.path.basename(video)

    if video_name in train_names:
        train_videos.append(video)
    elif video_name in test_names:
        test_videos.append(video)
    else:
        unmatched_videos.append(video)


print("Training videos:", len(train_videos))
print("Testing videos:", len(test_videos))
print("Unmatched videos:", len(unmatched_videos))
```

```python
import pandas as pd
import os

video_records = []

for video in train_videos:
    video_records.append({
        "video_path": video,
        "video_name": os.path.basename(video),
        "category": os.path.basename(os.path.dirname(video)),
        "split": "train",
        "label": 1
    })

for video in test_videos:
    video_records.append({
        "video_path": video,
        "video_name": os.path.basename(video),
        "category": os.path.basename(os.path.dirname(video)),
        "split": "test",
        "label": 1
    })

part1_df = pd.DataFrame(video_records)

print("Metadata shape:", part1_df.shape)
display(part1_df.head())
```
