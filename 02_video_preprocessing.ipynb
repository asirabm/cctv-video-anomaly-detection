**Download the official split files**

```python
from huggingface_hub import hf_hub_download

repo_id = "jinmang2/ucf_crime"
split_folder = "UCF_Crimes-Train-Test-Split/Anomaly_Detection_splits/"

# downloading train split
train_split_path = hf_hub_download(
    repo_id=repo_id,
    filename=split_folder + "Anomaly_Train.txt",
    repo_type="dataset"
)
# downloading test splits
test_split_path = hf_hub_download(
    repo_id=repo_id,
    filename=split_folder + "Anomaly_Test.txt",
    repo_type="dataset"
)

print("Training split:", train_split_path)
print("Testing split:", test_split_path)
```

**Identify training and testing videos**

```python

with open(train_split_path, "r") as f:
    train_lines = [line.strip() for line in f if line.strip()]

with open(test_split_path, "r") as f:
    test_lines = [line.strip() for line in f if line.strip()]


train_names = {line.split("/")[-1].replace("\\", "/") for line in train_lines}
test_names = {line.split("/")[-1].replace("\\", "/") for line in test_lines}


train_videos = []
test_videos = []
unmatched_videos = []

for video in extracted_videos:
    video_name = os.path.basename(video)

    if video_name in train_names:
        train_videos.append(video)
    elif video_name in test_names:
        test_videos.append(video)
    else:
        unmatched_videos.append(video)


print("Training videos:", len(train_videos))
print("Testing videos:", len(test_videos))
print("Unmatched videos:", len(unmatched_videos))
```

**Prepare Pre-Trained ResNet50 to extract information from frames**

```python
import torch
from torchvision import models, transforms


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")


weights = models.ResNet50_Weights.DEFAULT
model = models.resnet50(weights=weights)

# Removing the final classification layer
model = torch.nn.Sequential(*list(model.children())[:-1])

# Set the model to evaluation mode, becoz model is already trained
model = model.to(device)
model.eval()

# Image preprocessing required by the pretrained weights,
## these weights handles the required operations, including resizing, cropping, converting the image to a tensor, and normalising its values.
preprocess = weights.transforms()

print("ResNet50 is ready.")
print("Device:", device)
```

**Testing ResNet50 to extract information from a video's frame**

```python
import cv2
import numpy as np
import torch
from PIL import Image

# select the first video from traing set
video_path = train_videos[0]

# open the slecetd video
cap = cv2.VideoCapture(video_path)

# Read the first frame of the sopend video
success, frame = cap.read()
cap.release()

if success:
    # Convert OpenCV BGR image to RGB, bcoz OpenCV read images in BGR order
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    image = Image.fromarray(frame)

    # Apply ResNet50 preprocessing
    image_tensor = preprocess(image).unsqueeze(0).to(device)

    # Extract features
    with torch.no_grad():
        features = model(image_tensor)

    # Convert features to a NumPy array
    features = features.squeeze().cpu().numpy()

    print("Video:", video_path)
    print("Feature shape:", features.shape)
    print("Feature data type:", features.dtype)
else:
    print("Could not read the video frame.")
```

**Extracting features from 16 selected frames of one video.**

```python
import cv2
import numpy as np
import torch
from PIL import Image

video_path = train_videos[0]
num_frames = 16

cap = cv2.VideoCapture(video_path)
# get total frames of the video
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if total_frames > 0:
    # Select 16 frames acrosss the video
    frame_indices = np.linspace(
        0, total_frames - 1, num_frames, dtype=int
    )

    video_features = []
#process each frames
    for frame_index in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))
        success, frame = cap.read()

        if not success:
            continue

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image = Image.fromarray(frame)
        image_tensor = preprocess(image).unsqueeze(0).to(device)

        with torch.no_grad():
            features = model(image_tensor)

        features = features.squeeze().cpu().numpy()
        video_features.append(features)

cap.release()

video_features = np.array(video_features, dtype=np.float32)

print("Video:", video_path)
print("Total frames in video:", total_frames)
print("Frames successfully processed:", len(video_features))
print("Feature shape:", video_features.shape)
```

**Extract and save features for five videos for testing purpose**

```python
import os
import numpy as np
import cv2
import torch
from PIL import Image

feature_root = "/content/drive/MyDrive/UCF_Crime_Project/features"

num_frames = 16
results = []

# Test with the first five videos
for _, row in video_df.head(5).iterrows():

    video_path = row["video_path"]
    video_name = row["video_name"]
    category = row["category"]
    split = row["split"]

    # Create the correct folder automatically
    save_folder = os.path.join(feature_root, split, category)
    os.makedirs(save_folder, exist_ok=True)

    # Create the output filename
    feature_name = os.path.splitext(video_name)[0] + ".npy"
    feature_path = os.path.join(save_folder, feature_name)

    # Skip files that have already been saved correctly
    if os.path.exists(feature_path):
        saved_features = np.load(feature_path)

        if saved_features.shape == (num_frames, 2048):
            print("Already saved:", video_name)
            results.append({
                "video_name": video_name,
                "status": "Already saved"
            })
            continue

    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames < num_frames:
        cap.release()
        print("Not enough frames:", video_name)
        results.append({
            "video_name": video_name,
            "status": "Failed: not enough frames"
        })
        continue

    frame_indices = np.linspace(
        0, total_frames - 1, num_frames, dtype=int
    )

    features_list = []

    for frame_index in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))
        success, frame = cap.read()

        if not success:
            break

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image = Image.fromarray(frame)
        image_tensor = preprocess(image).unsqueeze(0).to(device)

        with torch.no_grad():
            features = model(image_tensor)

        features = features.squeeze().cpu().numpy()
        features_list.append(features)

    cap.release()

    if len(features_list) == num_frames:
        video_features = np.array(features_list, dtype=np.float32)
        np.save(feature_path, video_features)

        print("Saved:", video_name, video_features.shape)

        results.append({
            "video_name": video_name,
            "status": "Success"
        })

    else:
        print("Failed to process:", video_name)

        results.append({
            "video_name": video_name,
            "status": "Failed: frame reading"
        })

print("\nTest batch finished.")
```

**Process all 200 videos in Part 1**

```python

import os
import cv2
import numpy as np
import pandas as pd
import torch
from PIL import Image

feature_root = "/content/drive/MyDrive/UCF_Crime_Project/features"
log_path = "/content/drive/MyDrive/UCF_Crime_Project/metadata/part1_processing_log.csv"

num_frames = 16
results = []

total_videos = len(video_df)

for index, row in video_df.iterrows():

    video_path = row["video_path"]
    video_name = row["video_name"]
    category = row["category"]
    split = row["split"]

    save_folder = os.path.join(feature_root, split, category)
    os.makedirs(save_folder, exist_ok=True)

    feature_name = os.path.splitext(video_name)[0] + ".npy"
    feature_path = os.path.join(save_folder, feature_name)

    # Skip a video if its saved features are valid
    if os.path.exists(feature_path):
        try:
            saved_features = np.load(feature_path)

            if (saved_features.shape == (num_frames, 2048)
                    and np.isfinite(saved_features).all()):

                results.append({
                    "video_name": video_name,
                    "category": category,
                    "split": split,
                    "status": "Success"
                })

                print(f"[{index + 1}/{total_videos}] Already saved: {video_name}")
                continue

        except Exception:
            pass

    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if not cap.isOpened() or total_frames < num_frames:
        cap.release()

        results.append({
            "video_name": video_name,
            "category": category,
            "split": split,
            "status": "Failed: video unavailable or too short"
        })

        print(f"[{index + 1}/{total_videos}] Failed: {video_name}")
        continue

    frame_indices = np.linspace(
        0, total_frames - 1, num_frames, dtype=int
    )

    features_list = []

    for frame_index in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))
        success, frame = cap.read()

        if not success:
            break

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image = Image.fromarray(frame)
        image_tensor = preprocess(image).unsqueeze(0).to(device)

        with torch.no_grad():
            features = model(image_tensor)

        features = features.squeeze().cpu().numpy()
        features_list.append(features)

    cap.release()

    if len(features_list) == num_frames:
        video_features = np.array(features_list, dtype=np.float32)

        if np.isfinite(video_features).all():
            np.save(feature_path, video_features)

            status = "Success"
            print(
                f"[{index + 1}/{total_videos}] "
                f"Saved: {video_name} {video_features.shape}"
            )
        else:
            status = "Failed: invalid feature values"
            print(f"[{index + 1}/{total_videos}] Failed: {video_name}")
    else:
        status = "Failed: frame reading"
        print(f"[{index + 1}/{total_videos}] Failed: {video_name}")

    results.append({
        "video_name": video_name,
        "category": category,
        "split": split,
        "status": status
    })

    # Save the log after every video
    pd.DataFrame(results).to_csv(log_path, index=False)

print("\nPart 1 feature extraction finished.")
print(pd.DataFrame(results)["status"].value_counts())
print("Processing log:", log_path)
```

**verify the saved files**

```python

import os
import numpy as np
from collections import Counter

feature_root = "/content/drive/MyDrive/UCF_Crime_Project/features"

feature_counts = Counter()
invalid_files = []

for split in ["train", "test"]:
    split_folder = os.path.join(feature_root, split)

    for category in os.listdir(split_folder):
        category_folder = os.path.join(split_folder, category)

        if not os.path.isdir(category_folder):
            continue

        for file_name in os.listdir(category_folder):
            if file_name.endswith(".npy"):
                feature_counts[split] += 1

                file_path = os.path.join(category_folder, file_name)
                features = np.load(file_path)

                if (features.shape != (16, 2048)
                        or not np.isfinite(features).all()):
                    invalid_files.append(file_path)

print("Training feature files:", feature_counts["train"])
print("Testing feature files:", feature_counts["test"])
print("Total feature files:", sum(feature_counts.values()))
print("Invalid feature files:", len(invalid_files))
```
